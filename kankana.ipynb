```python
from collections import deque

capacity_a = 4
capacity_b = 3
goal = 2

def get_next_states(a, b):
    states = []

    states.append(((capacity_a, b), "Fill Jug A"))
    states.append(((a, capacity_b), "Fill Jug B"))
    states.append(((0, b), "Empty Jug A"))
    states.append(((a, 0), "Empty Jug B"))

    pour = min(a, capacity_b - b)
    states.append(((a - pour, b + pour), "Pour Jug A into Jug B"))

    pour = min(b, capacity_a - a)
    states.append(((a + pour, b - pour), "Pour Jug B into Jug A"))

    return states

def bfs():
    queue = deque()
    queue.append(((0, 0), []))
    visited = set()

    while queue:
        (a, b), path = queue.popleft()

        if (a, b) in visited:
            continue

        visited.add((a, b))

        if a == goal or b == goal:
            return path + [(a, b)]

        for state, operation in get_next_states(a, b):
            if state not in visited:
                queue.append((state, path + [(a, b, operation)]))

solution = bfs()

print("Minimum sequence of steps:")

for i, step in enumerate(solution):
    if len(step) == 3:
        a, b, operation = step
        print(f"Step {i}: ({a}, {b}) -> {operation}")
    else:
        a, b = step
        print(f"Step {i}: ({a}, {b}) -> Goal reached")
```
