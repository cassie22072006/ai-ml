```python
from collections import deque
import matplotlib.pyplot as plt

graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': [],
    'F': []
}

def bfs(start):
    visited = []
    queue = deque([start])

    while queue:
        node = queue.popleft()

        if node not in visited:
            visited.append(node)
            queue.extend(graph[node])

    return visited

def dfs(start):
    visited = []

    def visit(node):
        visited.append(node)

        for neighbour in graph[node]:
            if neighbour not in visited:
                visit(neighbour)

    visit(start)
    return visited

bfs_order = bfs('A')
dfs_order = dfs('A')

print("BFS Traversal:", " -> ".join(bfs_order))
print("DFS Traversal:", " -> ".join(dfs_order))

positions = {
    'A': (0, 2),
    'B': (-1, 1),
    'C': (1, 1),
    'D': (-1.5, 0),
    'E': (-0.5, 0),
    'F': (1, 0)
}

edges = [
    ('A', 'B'),
    ('A', 'C'),
    ('B', 'D'),
    ('B', 'E'),
    ('C', 'F')
]

def draw_tree(order, title):
    plt.figure(figsize=(7, 5))

    for parent, child in edges:
        x1, y1 = positions[parent]
        x2, y2 = positions[child]
        plt.plot([x1, x2], [y1, y2], 'k-')

    for i, node in enumerate(order, start=1):
        x, y = positions[node]
        plt.scatter(x, y, s=1000)
        plt.text(x, y, f"{node}\n({i})",
                 ha='center', va='center', fontsize=11)

    plt.title(title)
    plt.axis('off')
    plt.show()

draw_tree(bfs_order, "BFS Traversal Tree")
draw_tree(dfs_order, "DFS Traversal Tree")
```
