```python
graph = {
    'kolkata': ['durgapur', 'asansol'],
    'durgapur': ['kolkata', 'bankura'],
    'asansol': ['kolkata', 'bankura'],
    'bankura': ['durgapur', 'asansol'],
    'mumbai': ['pune'],
    'pune': ['mumbai'],
    'goa': []
}

def route_exists(graph, start, destination):
    visited = set()

    def dfs(city):
        if city == destination:
            return True

        visited.add(city)

        for neighbour in graph[city]:
            if neighbour not in visited:
                if dfs(neighbour):
                    return True

        return False

    return dfs(start)

def find_components(graph):
    visited = set()
    components = []

    def dfs(city, component):
        visited.add(city)
        component.append(city)

        for neighbour in graph[city]:
            if neighbour not in visited:
                dfs(neighbour, component)

    for city in graph:
        if city not in visited:
            component = []
            dfs(city, component)
            components.append(component)

    return components

start = 'kolkata'
destination = 'bankura'

if route_exists(graph, start, destination):
    print(f"Route exists from {start} to {destination}")
else:
    print(f"No route exists from {start} to {destination}")

components = find_components(graph)

print("\nConnected Components:")
for i, component in enumerate(components, start=1):
    print(f"Component {i}: {' -> '.join(component)}")
```
